```python
import torch
import torchvision
from torchvision.models.detection import fasterrcnn_resnet50_fpn, FasterRCNN_ResNet50_FPN_Weights

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

model = fasterrcnn_resnet50_fpn(weights=FasterRCNN_ResNet50_FPN_Weights.DEFAULT)
model.to(device)

images = [torch.rand(3, 300, 300).to(device), torch.rand(3, 300, 300).to(device)]
targets = [
    {
        'boxes': torch.tensor([[50.0, 50.0, 150.0, 150.0]], dtype=torch.float32, device=device),
        'labels': torch.tensor([1], dtype=torch.int64, device=device)
    },
    {
        'boxes': torch.tensor([[30.0, 40.0, 120.0, 180.0]], dtype=torch.float32, device=device),
        'labels': torch.tensor([2], dtype=torch.int64, device=device)
    }
]

optimizer = torch.optim.SGD(model.parameters(), lr=0.005, momentum=0.9, weight_decay=0.0005)

model.train()
epochs = 5
for epoch in range(epochs):
    optimizer.zero_grad()
    loss_dict = model(images, targets)
    losses = sum(loss for loss in loss_dict.values())
    losses.backward()
    optimizer.step()
    print(f"Epoch [{epoch+1}/{epochs}], Loss: {losses.item():.4f}")

model.eval()
with torch.no_grad():
    predictions = model([images[0]])
    print("\nPredicted Boxes:\n", predictions[0]['boxes'][:2].cpu().numpy())
    print("Predicted Scores:\n", predictions[0]['scores'][:2].cpu().numpy())
    print("Predicted Labels:\n", predictions[0]['labels'][:2].cpu().numpy())
```
